**A** <br>
Vi antar at $M$ ikke er inverterbar ($Mx=0$ har flere løsninger enn $x=0$)
![2a.png](2a.png)

<br> Vi har kun en løsning fordi: <br>
![1_losning.png](1_losning.png) <br>
Vi har at for en stokoastisk kolonne at alle eigenverdier har absoluttv verdi høyest 1. Vi har vist at hvis $M$ ikke er inverterbar vil $S$ ha en eigenverdi større enn 1. Dette er en motsigelse ($|\lambda| \leq 1$ kan ikke stemme samtidig som $\lambda = \frac{1}{a}>1$), så $M$ er inverterbar. $Mp*=b$ må derfor ha bare en løsning, $p*=M^{-1}b$.  <br>

**C**  <br>
Start fordelingen kan ikke endre den stasjonære løsningen fordi $p*=M^{-1}b$ der M og b er uavhengig fra start fordelingen. <br>
Med verdiene vår har vi at feilen mot direkte løsning er rundt 10^-9, er avviket i p forhold til  p* er svært liten. Beggestart verdiene gir samme løsning, dermed har vi bekreftet at det bare er en løsning <br>
![rangering.png](rangering.png)

Enighet mellom interasjon og direkte løsning kontrollerer at begge metdoene regner ut riktig fikspunkt for T, med de gitte parameterne u, a og S. Begge metodene kan være enige fordi de begge bruker S direkte. Enigheten mellom iterasjon og direkte løsning kan finne feil i regningen, men ikke feil i S. <br>

Vi sammenlikner residualnormen med $(1-\alpha)\cdot 10^{-8}$ fordi vi ikke har $p^*$ direkte, så vi bruker residualen i stedet. Siden feilen kan være opptil residualen delt på $(1-\alpha)$, gir dette $\|p-p^*\|_1 \leq \frac{\|T(p)-p\|_1}{1-\alpha}$, som gir $(1-\alpha)\|p-p^*\|_1 \leq \|T(p)-p\|_1$. Koden har toleranse $(1-\alpha)\cdot 10^{-8}$ og stopper bare når residualen er høyst denne. Med $\alpha = 0.85$ får vi da $(1-0.85)\|p-p^*\|_1 \leq \|T(p)-p\|_1 \leq (1-0.85)\cdot 10^{-8}$, så feilen er høyst $10^{-8}$.
