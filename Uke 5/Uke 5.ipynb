Vi antar at C vil få høyest verdi, fordi C har en inn sum på 1.5 og kun sender hele sin vekt til A. Den får så tilbake 1/2 av dette fra A, den får også 1/4 av sin vekt tilbake fra syklusen som A,B, C danner. Den får også 1/8 fra av sin vekt (C) fra D. <br>
D som også har en inn sum på 1.5 sender, men den motar 1/4 av C, så sender den 1/2 av sin vekt til C og resten i en syklus D, E, F. Dette er 1/2 av D, som deretter blir splittet med syklusen og C. Den får 1/4 og sender ut 1/2, så den vil minke.
